```python
#Initialise the drawing board
#Sets the width to 7 inches and the height to 5 inches. If you want a wider or taller graph, you change these numbers.
plt.figure(figsize=(7,5))
#Create the scatter plot
#Styling - color of the points, marker * instead of dots, size represented by s (default = 20), alpha = transparency
plt.scatter(x=df.study_hours,y=df.student_marks,color="blue",marker="*",s=100, alpha = 0.45)
plt.show()
```
